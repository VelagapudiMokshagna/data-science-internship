```python
import pandas as pd
```

```python
df = pd.read_csv("twitter_training.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.columns = ["ID", "Entity", "Sentiment", "Tweet"]
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=["Tweet"])
```

```python
df.shape
```

```python
df["Sentiment"].value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
sentiment_counts = df["Sentiment"].value_counts()

sentiment_counts.plot(kind="bar")

plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Tweets")

plt.show()
```

```python
df["Entity"].value_counts().head(10)
```

```python
entity_counts = df["Entity"].value_counts().head(10)

entity_counts.plot(kind="bar")

plt.title("Top 10 Most Discussed Entities")
plt.xlabel("Entity")
plt.ylabel("Number of Tweets")
plt.xticks(rotation=45)

plt.show()
```

```python
sentiment_entity = pd.crosstab(df["Entity"], df["Sentiment"])

sentiment_entity.head()
```

```python
sentiment_entity.plot(kind="bar", stacked=True, figsize=(12, 6))

plt.title("Sentiment Distribution by Entity")
plt.xlabel("Entity")
plt.ylabel("Number of Tweets")
plt.xticks(rotation=90)

plt.show()
```

## Conclusion

- Negative sentiment was the most common sentiment in the dataset.
- Positive sentiment was the second most common, followed by Neutral sentiment.
- Irrelevant tweets represented the smallest category.
- Sentiment patterns vary across different entities/topics.
- The analysis provides an overview of public opinion toward different topics and brands.
